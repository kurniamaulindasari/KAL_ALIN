# SINGULAR VALUE DECOMPOSITION (SVD)

## Apa itu SVD
SVD (Singular Value Decomposition) adalah salah satu teknik dekomposisi matriks yang sangat penting dalam aljabar linear dan banyak digunakan dalam ilmu komputer, statistik, dan machine learning.
SVD memecah sebuah matriks persegi panjang (bukan hanya matriks bujur sangkar) menjadi tiga matriks:


$$
A = U \Sigma V^T
$$

### Penjelasan komponen:

* **A**: Matriks asli berukuran $m \times n$
* **U**: Matriks ortogonal berukuran $m \times m \rightarrow$ kolomnya disebut *left singular vectors*
* **$\Sigma$** (Sigma): Matriks diagonal $m \times n$ berisi *singular values* (nilai-nilai yang non-negatif dan biasanya diurutkan dari besar ke kecil)
* **V$^\text{T}$**: Transpose dari matriks ortogonal $n \times n \rightarrow$ barisnya adalah *right singular vectors*

SVD dapat diterapkan pada semua matriks, bahkan jika matriks tidak memiliki invers.

### Kegunaan SVD
SVD sangat banyak digunakan dalam berbagai bidang. Berikut adalah kegunaan utamanya:

🔍 Reduksi Dimensi (Dimensionality Reduction)
Digunakan dalam PCA (Principal Component Analysis) untuk mengurangi jumlah variabel dalam dataset tanpa kehilangan informasi penting.

Contoh: Dalam pengolahan citra atau teks, SVD mengurangi ukuran data untuk menghemat ruang penyimpanan dan mempercepat proses analisis.

🧠 Machine Learning & Data Mining
Pengenalan Pola: Untuk ekstraksi fitur atau representasi data yang lebih kompak.

Rekomendasi Sistem: Misalnya dalam Netflix atau YouTube, SVD digunakan untuk memprediksi rating berdasarkan pola yang tersembunyi dalam data pengguna dan produk.

📊 Pemrosesan Citra (Image Compression)
SVD dapat digunakan untuk menyimpan gambar dengan ukuran lebih kecil tetapi tetap mempertahankan kualitas visual yang baik.

Prinsipnya: Simpan hanya nilai singular terbesar dan buang yang kecil (yang menyumbang sedikit informasi).

🧮 Solusi Persamaan Linear
Untuk menyelesaikan sistem persamaan linear yang tidak memiliki solusi eksak (overdetermined atau underdetermined system).

Juga digunakan dalam pseudoinverses untuk mencari solusi terbaik (least-squares solution).

📈 Penyaring Noise / Denoising
Dalam sinyal atau gambar, SVD bisa memisahkan komponen utama dari noise.

Komponen dengan singular value kecil sering dianggap sebagai noise dan bisa dibuang.

📘 Natural Language Processing (NLP)
Digunakan dalam Latent Semantic Analysis (LSA) untuk memodelkan hubungan antara dokumen dan kata-kata secara matematis.

SVD membantu menemukan makna tersembunyi di balik teks.



### Bentuk Formula SVD
Jika $𝐴$ adalah $𝑚×𝑛$, maka:

$$
A = U \Sigma V^T =
\begin{bmatrix}
\,|\, & \,|\, &        & \,|\, \\
u_1 & u_2 & \cdots & u_m \\
\,|\, & \,|\, &        & \,|\, \\
\end{bmatrix}
\begin{bmatrix}
\sigma_1 & 0        & \cdots & 0 \\
0        & \sigma_2 & \cdots & 0 \\
\vdots   & \vdots   & \ddots & \vdots \\
0        & 0        & \cdots & \sigma_r \\
\end{bmatrix}
\begin{bmatrix}
- & v_1^T & - \\
- & v_2^T & - \\
  & \vdots &   \\
- & v_n^T & - \\
\end{bmatrix}
$$


```python
# code pembuktian
import numpy as np

# Matriks A dari soal
A = np.array([
    [3, 7],
    [2, 5],
    [4, 3],
    [1, 1]
], dtype=float)

# Langkah 1: Hitung A^T A
A_TA = A.T @ A
print("A^T A =\n", A_TA)

# Langkah 2: Hitung eigenvalue dan eigenvector A^T A
eigvals, eigvecs = np.linalg.eigh(A_TA)

# Urutkan dari terbesar ke terkecil
idx = np.argsort(eigvals)[::-1]
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]

print("\nEigenvalues (A^T A) =", eigvals)
print("Eigenvectors (V) =\n", eigvecs)

# Langkah 3: Hitung singular value
singular_values = np.sqrt(eigvals)
print("\nSingular values =", singular_values)

# Bentuk Sigma (4x2 karena A 4x2)
Sigma = np.zeros((4, 2))
Sigma[0, 0] = singular_values[0]
Sigma[1, 1] = singular_values[1]
print("\nSigma =\n", Sigma)

# Langkah 4: Matriks V
V = eigvecs

# Langkah 5: Hitung U = (1/σ) * A * v
u1 = (A @ V[:, 0]) / singular_values[0]
u2 = (A @ V[:, 1]) / singular_values[1]

# Gabungkan jadi matriks U (4x2 untuk dua singular value)
U = np.column_stack((u1, u2))
print("\nU (2 kolom pertama) =\n", U)

# Langkah 6: Buktikan A = U Σ V^T
A_reconstructed = U @ Sigma[:2, :] @ V.T
print("\nA reconstructed =\n", A_reconstructed)

print("\nA original =\n", A)

# Hitung error rekonstruksi
error = np.linalg.norm(A - A_reconstructed)
print(f"\nReconstruction error = {error}")
```
