# MATRIKS

## DEFINISI MATRIKS
Matriks adalah struktur matematika yang terdiri dari sekumpulan bilangan, simbol, atau ekspresi yang disusun dalam baris (rows) dan kolom (columns). Matriks digunakan untuk merepresentasikan data dalam bentuk tabel dua dimensi, di mana setiap elemen matriks memiliki posisi yang unik berdasarkan indeks baris dan kolomnya.

### OPERASI ARITMATIKA MATRIKS
Operasi aritmatika matriks adalah serangkaian operasi matematis yang dilakukan pada matriks. Operasi ini melibatkan penjumlahan, pengurangan, perkalian, dan operasi lainnya yang memanipulasi elemen-elemen matriks.

#### PENJUMLAHAN MATRIKS
Penjumlahan matriks dilakukan dengan menjumlahkan elemen-elemen yang berada pada posisi yang sama dari dua matriks. Syaratnya adalah kedua matriks harus memiliki dimensi yang sama (jumlah baris dan kolom yang sama).

##### Rumus Penjumlahan Matriks
Jika $A$ dan $B$ adalah dua matriks berukuran $m$ x $n$, maka hasil penjumlahan $C = A + B$ adalah matriks $m$ x $n$ dengan elemen :

$$
\begin{array}{cc}
c_{ij}&=a_{ij}+b_{ij}\\
\end{array}
$$

dimana $i$ adalah indeks baris dan $j$ adalah indeks kolom.

##### Contoh Penjumlahan Matriks
$A$ =
$
\begin{bmatrix}
1 & 2 \\
3 & 4 \\
\end{bmatrix}
\quad
$

$B$ =
$$
\begin{bmatrix}
5 & 6 \\
7 & 8 \\
\end{bmatrix}
\quad
$$

$A+B$ =

$$
\begin{bmatrix}
1 + 5 & 2 + 6\\
3 + 7 & 4 + 8\\
\end{bmatrix}
\quad = \begin{bmatrix}
6&8\\
10&12\\
\end{bmatrix}
\quad
$$

```python
matriks_a = []
matriks_b = []
matriks_hasil = []

print("Matriks A")
for i in range(2):
    matriks_a_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks_a.append([int(angka) for angka in matriks_a_input.split()])

print()

print("Matriks B")
for i in range(2):
    matriks_b_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks_b.append([int(angka) for angka in matriks_b_input.split()])

print("\nTampilan matriks A:")
for baris in matriks_a:
    print(baris)

print("\nTampilan matriks B:")
for baris in matriks_b:
    print(baris)

matriks_hasil = [[0 for _ in range(2)] for _ in range(2)]

for i in range(2):
    for j in range(2):
        matriks_hasil[i][j] = matriks_a[i][j] + matriks_b[i][j]

print("\nHasil Penjumlahan Matriks A + B :")
for baris in matriks_hasil:
    print(baris)
```

#### PENGURANGAN MATRIKS
Pengurangan matriks mirip dengan penjumlahan, tetapi elemen-elemennya dikurangkan. Syaratnya sama: kedua matriks harus memiliki dimensi yang sama.

##### Rumus Pengurangan Matriks
Jika $A$ dan $B$ adalah dua matriks berukuran $m$ x $n$, maka hasil pengurangan $C = A - B$ adalah matriks berukuran $m$ x $n$ dengan elemen:

$$
\begin{array}{cc}
c_{ij}&=a_{ij}-b_{ij}\\
\end{array}
$$

##### Contoh Pengurangan Matriks
$A$ =
$$
\begin{bmatrix}
5 & 6 \\
7 & 8 \\
\end{bmatrix}
\quad
$$

$B$ =
$$
\begin{bmatrix}
1 & 2 \\
3 & 4 \\
\end{bmatrix}
\quad
$$

$A-B$ =

$$
\begin{bmatrix}
5 - 1 & 6 - 2\\
7 - 3 & 8 - 4\\
\end{bmatrix}
\quad = \begin{bmatrix}
4&4\\
4&4\\
\end{bmatrix}
\quad
$$

```python
matriks_a = []
matriks_b = []
matriks_hasil = []

print("Matriks A")
for i in range(2):
    matriks_a_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks_a.append([int(angka) for angka in matriks_a_input.split()])

print()

print("Matriks B")
for i in range(2):
    matriks_b_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks_b.append([int(angka) for angka in matriks_b_input.split()])

print("\nTampilan matriks A:")
for baris in matriks_a:
    print(baris)

print("\nTampilan matriks B:")
for baris in matriks_b:
    print(baris)

matriks_hasil = [[0 for _ in range(2)] for _ in range(2)]

for i in range(2):
    for j in range(2):
        matriks_hasil[i][j] = matriks_a[i][j] - matriks_b[i][j]

print("\nHasil Pengurangan Matriks A - B:")
for baris in matriks_hasil:
    print(baris)
```

#### PERKALIAN MATRIKS
Perkalian matriks adalah operasi yang lebih kompleks dibandingkan penjumlahan dan pengurangan. Perkalian matriks $A$ dan $B$ menghasilakn matriks baru $C$, dimana setiap elemen $c_{ij}$ dihiitung sebagai *dot product* dari baris ke-$i$ matriks $A$ dan kolom ke-$j$ matriks $B$.

##### Syarat Perkalian Matriks


*   Jumlah kolom matriks $A$ harus sama dengan jumlah baris matriks $B$.
*   Jika $A$ berukuran $m$ x $n$, maka $B$ harus berukuran $n$ x $p$.
*   Hasil perkalian $C$ akan berukuran $m$ x $p$.

##### Contoh Perkalian matriks
$A$ =
$$
\begin{bmatrix}
1 & 2 \\
3 & 4 \\
\end{bmatrix}
\quad
$$

$B$ =
$$
\begin{bmatrix}
5 & 6 \\
7 & 8 \\
\end{bmatrix}
\quad
$$

$C$ = $A$ x $B$ =
\begin{bmatrix}
(1 \cdot 5 + 2 \cdot 7) & (1 \cdot 6 + 2 \cdot 8) \\ (3 \cdot 5 + 4 \cdot 7) & (3 \cdot 6 + 4 \cdot 8) \end{bmatrix}
= \begin{bmatrix} 19 & 22 \\ 43 & 50 \end{bmatrix}

```python
matriks_a = []
matriks_b = []
matriks_hasil = []

print("Matriks A")
for i in range(2):
    matriks_a_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks_a.append([int(angka) for angka in matriks_a_input.split()])

print()

print("Matriks B")
for i in range(2):
    matriks_b_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks_b.append([int(angka) for angka in matriks_b_input.split()])

print("\nTampilan matriks A:")
for baris in matriks_a:
    print(baris)

print("\nTampilan matriks B:")
for baris in matriks_b:
    print(baris)

matriks_hasil = [[0 for _ in range(2)] for _ in range(2)]

for i in range(2):
    for j in range(2):
        for k in range(2):
            matriks_hasil[i][j] += matriks_a[i][k] * matriks_b[k][j]

print("\nHasil Perkalian Matriks A x B:")
for baris in matriks_hasil:
    print(baris)
```

#### PERKALIAN SKALAR DENGAN MATRIKS
Perkalian skalar melibatkan mengalikan setiap elemen matriks dengan sebuah bilangan skalar.

##### Rumus Perkalian Skalar
Jika $k$ adalah skalar dan $A$ adalah matriks, maka hasil perkalian $B$ = $k$ . $A$ adalah:
$$
\begin{array}{cc}
b_{ij}&=k.a_{ij}\\
\end{array}
$$

##### Contoh perkalian Skalar
$k = 2,$ $A =$
 \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}

$k \cdot A = $
\begin{bmatrix} 2 \cdot 1 & 2 \cdot 2 \\ 2 \cdot 3 & 2 \cdot 4 \end{bmatrix} = \begin{bmatrix} 2 & 4 \\ 6 & 8 \end{bmatrix}

```python
matriks = []
matriks_hasil = []

print("Masukkan elemen matriks (2x2):")
for i in range(2):
    matriks_input = input(f"Baris ke {i+1} (2 angka, dipisahkan dengan 1 spasi): ")
    matriks.append([int(angka) for angka in matriks_input.split()])

skalar = int(input("\nMasukkan bilangan skalar: "))

print("\nTampilan matriks asli:")
for baris in matriks:
    print(baris)

matriks_hasil = [[0 for _ in range(2)] for _ in range(2)]

for i in range(2):
    for j in range(2):
        matriks_hasil[i][j] = skalar * matriks[i][j]

print(f"\nHasil Perkalian Skalar {skalar} dengan Matriks:")
for baris in matriks_hasil:
    print(baris)
```
