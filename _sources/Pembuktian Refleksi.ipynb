```python
import numpy as np
import matplotlib.pyplot as plt

# Titik-titik objek asli berbentuk "V"
points = np.array([
    [1.5, 1.5],
    [2, 1],
    [1.5, 0.5],
    [1, 1]
]).T  # Bentuk 2xN

# Matriks refleksi terhadap garis y = x (tukar x dan y)
reflection_matrix = np.array([
    [0, 1],
    [1, 0]
])

# Hasil refleksi
reflected = reflection_matrix @ points

# Plotting
fig, ax = plt.subplots()

# Gambar bentuk asal (merah) dan refleksi (biru)
ax.plot(points[0], points[1], 'ro-', label='Asal')
ax.plot(reflected[0], reflected[1], 'bo-', label='Refleksi')

# Garis cermin y = x
x_line = np.linspace(0, 3, 100)
ax.plot(x_line, x_line, 'g--', label='y = x')

# Sumbu dan grid
ax.axhline(y=0, color='k', linestyle=':')
ax.axvline(x=0, color='k', linestyle=':')
ax.set_aspect('equal')
ax.set_xlim(0, 3)
ax.set_ylim(0, 3)
ax.grid(True)
ax.legend()
ax.set_title('Refleksi Bentuk "V" terhadap Garis y = x')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Titik-titik bentuk "V"
points = np.array([
    [1.5, 1.5],
    [1, 2],
    [0.5, 1.5],
    [1, 1]
]).T

# Matriks refleksi terhadap y = -x
reflection_matrix = np.array([
    [0, -1],
    [-1, 0]
])

# Hasil refleksi
reflected = reflection_matrix @ points

# Plotting
fig, ax = plt.subplots()

# Plot bentuk asli dan hasil refleksi
ax.plot(points[0], points[1], 'ro-', label='Asal')
ax.plot(reflected[0], reflected[1], 'bo-', label='Refleksi')

# Garis cermin y = -x
x_line = np.linspace(-3, 3, 100)
ax.plot(x_line, -x_line, 'g--', label='y = -x')

# Sumbu dan grid
ax.axhline(0, color='k', linestyle=':')
ax.axvline(0, color='k', linestyle=':')
ax.set_aspect('equal')
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.grid(True)
ax.legend()
ax.set_title('Refleksi Bentuk "V" terhadap Garis y = -x')
plt.show()
```

## Tugas
#### Titik awal : (2, 1)
#### T1 = Refleksi sumbu x
#### T2 = Refleksi sumbu y
#### T3 = Rotasi 90

```python
import numpy as np
import matplotlib.pyplot as plt

points = np.array([
    [1.5, 1.5],
    [2, 1],
    [1.5, 0.5],
    [1, 1]
]).T

ref_y_eq_x = np.array([[0, 1], [1, 0]])
ref_y_eq_negx = np.array([[0, -1], [-1, 0]])

hasil = ref_y_eq_negx @ ref_y_eq_x
composed = hasil @ points

fig, ax = plt.subplots()

ax.plot(points[0], points[1], 'ro-', label='Asal')

ax.plot(composed[0], composed[1], 'mo-', label='Refleksi y=x lalu y=-x')

x_vals = np.linspace(-3, 3, 100)
ax.plot(x_vals, x_vals, 'g--', label='y = x')
ax.plot(x_vals, -x_vals, 'b--', label='y = -x')

# Format sumbu dan tampilan
ax.axhline(0, color='k', linestyle=':')
ax.axvline(0, color='k', linestyle=':')
ax.set_aspect('equal')
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.grid(True)
ax.set_title('Komposisi Refleksi: y = x lalu y = -x')
ax.legend()
plt.show()
```

```python
%matplotlib inline

from math import pi, sin, cos

import matplotlib.pyplot as plt
import numpy as np

# Titik-titik awal objek (sesuai dengan objek awal berwarna merah pada gambar)
titik_asal = np.array([[1, 1], [1.5, 0.5], [2.5, 0.5], [2, 1], [1, 1]]).T

x_asal = titik_asal[0, :]
y_asal = titik_asal[1, :]

# Sudut rotasi (90 derajat atau pi/2 radian)
theta = pi / 2

# Matriks rotasi 90 derajat berlawanan arah jarum jam
R = np.array([[cos(theta), -sin(theta)],
              [sin(theta), cos(theta)]])

# Menerapkan rotasi pada titik-titik asal
titik_rotasi = R @ titik_asal
x_rotasi = titik_rotasi[0, :]
y_rotasi = titik_rotasi[1, :]

# Membuat gambar
fig, ax = plt.subplots()

# Plotting titik-titik asal (merah)
ax.plot(x_asal, y_asal, 'ro', label='Asal')
ax.plot(x_asal, y_asal, 'r--')

# Plotting titik-titik hasil rotasi (biru)
ax.plot(x_rotasi, y_rotasi, 'bo', label='Rotasi 90°')
ax.plot(x_rotasi, y_rotasi, 'b')

# Menggambar sumbu-sumbu koordinat
ax.axvline(x=0, color="k", ls=":")
ax.axhline(y=0, color="k", ls=":")

# Menggambar garis y = x dan y = -x (sesuai gambar)
x_garis = np.linspace(-3, 3, 100)
ax.plot(x_garis, x_garis, 'g--', label='y = x')
ax.plot(x_garis, -x_garis, 'b--', label='y = -x')

# Konfigurasi plot
ax.grid(True)
ax.axis([-3, 3, -3, 3])
ax.set_aspect('equal')
ax.set_title("Rotasi 90° (Setara dengan Komposisi Refleksi)")
ax.legend()
plt.show()
```
