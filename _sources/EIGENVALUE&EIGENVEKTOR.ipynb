# EIGENVALUE & EIGENVEKTOR

## 1. EigenValue (Nilai Eigen)
### 📌 Pengertian Eigenvalue
Eigenvalue adalah skalar (bilangan) khusus yang muncul dalam transformasi linier. Secara intuitif, eigenvalue menunjukkan seberapa besar vektor eigen diperbesar atau diperkecil saat ditransformasi oleh sebuah matriks.

Jika diberikan sebuah matriks kuadrat $A$ dan sebuah vektor tak-nol $V$ , maka $V$ disebut eigenvector dari $A$ jika:

$$
A\vec{v} = \lambda \vec{v}
$$

Di sini:
* $A$ = matriks transformasi (ukuran $n \times n$)
* $\vec{v} \neq \vec{0}$ = eigenvector
* $\lambda$ = eigenvalue

### Tujuan Eigenvalue


> Untuk menganalisis bagaimana sebuah transformasi linier mengubah arah dan skala vektor.


> Digunakan dalam analisis getaran, kompresi data (PCA), mekanika kuantum, algoritma Google PageRank, dll.

### Cara Menentukan EigenValue

$$
A\vec{v} = \lambda\vec{v} \implies A\vec{v} - \lambda\vec{v} = \vec{0} \implies (A - \lambda I)\vec{v} = \vec{0}
$$

Untuk $\vec{v} \neq \vec{0}$, sistem ini memiliki solusi jika dan hanya jika determinan dari

$$
(A - \lambda I) = 0
$$

adalah nol :

> $$
\det(A - \lambda I) = 0
$$

Ini disebut persamaan karakteristik, dan solusinya $\lambda$ adalah eigenvalue.

## 2. EigenVektor
### 📌 Pengertian Eigenvektor
Eigenvektor adalah vektor tak nol yang arahnya tidak berubah ketika dikenai suatu transformasi linier oleh matriks.
Secara matematis, jika diberikan suatu matriks persegi $A$, maka vektor $V$ adalah eigenvektor dari $A$ jika memenuhi:

$$
A{v} = \lambda{v}
$$

Dimana:
* $v$ adalah eigenvektor
* $\lambda$ adalah eigenvalue (nilai eigen)
* ${v} \neq{0}$ (tidak boleh vektor nol)

### Ciri-ciri Eigenvektor
> Tidak boleh nol $({v} \neq{0})$

> Arah vektor tetap setelah dikalikan dengan matriks

> Eigenvektor hanya bisa eksis untuk matriks persegi

> Untuk setiap eigenvalue, biasanya ada satu atau lebih eigenvektor yang menyertainya





# Tabel Perbandingan Eigenvalue dan Eigenvector

---

| Aspek                   | Eigenvalue (Nilai Eigen)                       | Eigenvektor (Vektor Eigen)                        |
| :---------------------- | :--------------------------------------------- | :------------------------------------------------ |
| **Definisi** | Skalar $\lambda$ yang memenuhi $A\vec{v} = \lambda\vec{v}$ | Vektor tak nol $\vec{v}$ yang memenuhi $A\vec{v} = \lambda\vec{v}$ |
| **Jenis** | Bilangan real atau kompleks                    | Vektor berdimensi $n$, bisa real atau kompleks    |
| **Bentuk** | Angka/skalar                                   | Vektor kolom (misal $\begin{pmatrix} x \\ y \end{pmatrix}$) |
| **Fungsi dalam transformasi** | Menyatakan skala perubahan panjang vektor    | Menyatakan arah yang tidak berubah oleh transformasi matriks $A$ |
| **Simbol** | $\lambda$                                      | $\vec{v}$ atau $\mathbf{v}$                       |
| **Syarat** | $\det(A - \lambda I) = 0$                      | $(A - \lambda I)\vec{v} = \vec{0}$               |
| **Sifat penting** | Dapat memiliki beberapa nilai (dengan multiplisitas) | Bisa ada banyak vektor untuk satu nilai eigen       |
| **Keterkaitan** | Tiap eigenvalue berkaitan dengan satu atau lebih eigenvector | Tiap eigenvector selalu berpasangan dengan satu eigenvalue |
| **Tidak Unik?** | Nilai eigen unik (kecuali berulang)            | Vektor eigen tidak unik; dapat dikalikan dengan skalar $\neq 0$ |
| **Aplikasi** | Reduksi dimensi, stabilitas sistem, getaran mekanis | Arah deformasi tetap, analisis bentuk, machine learning (PCA), graf |
