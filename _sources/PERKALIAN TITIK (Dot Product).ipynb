# PERKALIAN TITIK (Dot Product)


## Dot product (produk titik) adalah operasi matematika yang dilakukan antara dua vektor dalam ruang berdimensi-𝑛, dan hasil dari operasi ini adalah sebuah skalar (bukan vektor).

### Notasi:
Jika ada dua vektor:

$$
\vec{a} = \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{bmatrix}, \quad \vec{b} = \begin{bmatrix} b_1 \\ b_2 \\ \vdots \\ b_n \end{bmatrix}
$$

maka dot product dari $\vec{a} \cdot \vec{b}$ adalah:

$$
\vec{a} \cdot \vec{b} = a_1b_1 + a_2b_2 + \dots + a_nb_n = \sum_{i=1}^{n} a_ib_i
$$

### Konsep Geometris Dot Product

Secara geometris, dot product antara dua vektor dapat dihitung dengan rumus:

$$
\vec{a} \cdot \vec{b} = |\vec{a}| |\vec{b}| \cos \theta
$$

di mana:

* $|\vec{a}|$ adalah panjang (norma) dari vektor $\vec{a}$

* $|\vec{b}|$ adalah panjang dari vektor $\vec{b}$

* $\theta$ adalah sudut antara vektor $\vec{a}$ dan $\vec{b}$

#### Interpretasi Geometris:

* Jika $\theta = 0^\circ$, maka $\cos \theta = 1 \rightarrow$ hasil maksimal (vektor searah).
* Jika $\theta = 90^\circ$, maka $\cos \theta = 0 \rightarrow$ hasil $= 0$ (vektor ortogonal / tegak lurus).
* Jika $\theta = 180^\circ$, maka $\cos \theta = -1 \rightarrow$ hasil negatif (vektor berlawanan arah).

### Contoh Perhitungan

#### Contoh 1: Dot Product Secara Aljabar

Misal:

$$
\vec{a} = \begin{bmatrix} 1 \\ 3 \\ -5 \end{bmatrix}, \quad \vec{b} = \begin{bmatrix} 4 \\ -2 \\ -1 \end{bmatrix}
$$


$$
\vec{a} \cdot \vec{b} = (1)(4) + (3)(-2) + (-5)(-1) = 4 - 6 + 5 = 3
$$


#### Contoh 2: Dot Product Secara Geometris

Misal dua vektor dengan panjang:

$$
|\vec{a}| = 5, \quad |\vec{b}| = 3, \quad \theta = 60^\circ
$$

$$
\vec{a} \cdot \vec{b} = 5 \cdot 3 \cdot \cos(60^\circ) = 15 \cdot 0.5 = 7.5
$$

### Sifat-sifat Dot Product

| Sifat                   | Penjelasan                                                                          |
| ----------------------- | ----------------------------------------------------------------------------------- |
| Komutatif               | $\vec{a} \cdot \vec{b} = \vec{b} \cdot \vec{a}$                                     |
| Distributif             | $\vec{a} \cdot (\vec{b} + \vec{c}) = \vec{a} \cdot \vec{b} + \vec{a} \cdot \vec{c}$ |
| Asosiatif dengan skalar | $(k \vec{a}) \cdot \vec{b} = k (\vec{a} \cdot \vec{b})$                             |
| Identitas               | $(\vec{a} \cdot \vec{a} = \vec{a}^2)$                                               |
| Ortogonalitas           | $\vec{a} \cdot \vec{b} = 0 \Rightarrow \vec{a} \perp \vec{b}$                       |


```python
def dot_product(v1, v2):
    if len(v1) != len(v2):
        raise ValueError("Panjang kedua vektor harus sama")

    result = sum(a * b for a, b in zip(v1, v2))
    return result

vec1 = [2, -3]
vec2 = [4, 1]

hasil = dot_product(vec1, vec2)
print("Dot product:", hasil)
```

# VEKTOR SATUAN

## Vektor satuan adalah vektor yang memiliki panjang (magnitudo) sebesar 1 satuan. Vektor ini digunakan untuk menunjukkan arah suatu vektor tanpa memperhatikan besarnya (panjangnya).

### Ciri-ciri Vektor Satuan:
* Panjangnya selalu 1

$$
|\vec{u}| = 1
$$

* Hanya menunjukkan arah, bukan besar gaya atau kecepatan.
* Biasa digunakan untuk membuat vektor lain dalam arah tertentu, misalnya saat melakukan normalisasi.

### Cara Mencari Vektor Satuan

Misalkan ada vektor sembarang:

$$
\vec{v} = \langle x, y \rangle
$$

Langkah-langkah untuk mencari vektor satuan dari
$\vec{v}$:

1.  Hitung panjang (magnitudo) vektor $\vec{v}$:

    $$
    |\vec{v}| = \sqrt{x^2 + y^2}
    $$

2.  Bagi setiap komponen vektor dengan panjangnya:

    $$
    \vec{u} = \frac{\vec{v}}{|\vec{v}|} = \left\langle \frac{x}{|\vec{v}|}, \frac{y}{|\vec{v}|} \right\rangle
    $$

> Vektor $\vec{u}$ inilah vektor satuan dalam arah yang sama dengan $\vec{v}$.

### Contoh:

Misalkan:

$$
\vec{v} = \langle 3,4 \rangle
$$

1.  Hitung panjang $\vec{v}$:

    $$
    |\vec{v}| = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5
    $$

2.  Vektor satuannya:

    $$
    \vec{u} = \frac{\vec{v}}{|\vec{v}|} = \left\langle \frac{3}{5}, \frac{4}{5} \right\rangle
    $$

Cek panjang $\vec{u}$:

$$
|\vec{u}| = \sqrt{\left(\frac{3}{5}\right)^2 + \left(\frac{4}{5}\right)^2} = \sqrt{\frac{9}{25} + \frac{16}{25}} = \sqrt{\frac{25}{25}} = \sqrt{1} = 1
$$

> $\checkmark$ Terbukti, $\vec{u}$ adalah vektor satuan.

```python
import math

def hitung_panjang_vektor(v):
    return math.sqrt(sum([x**2 for x in v]))

def vektor_satuan(v):
    panjang = hitung_panjang_vektor(v)
    if panjang == 0:
        raise ValueError("Vektor nol tidak bisa dikonversi ke vektor satuan.")
    return [x / panjang for x in v]

# Contoh penggunaan:
x = [3, 4]
u = vektor_satuan(x)

print("Vektor satuan:", u)

# Verifikasi panjang vektor satuan
panjang_u = hitung_panjang_vektor(u)
print("Panjang vektor satuan:", panjang_u)
```
